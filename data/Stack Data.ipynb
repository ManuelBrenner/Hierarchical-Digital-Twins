```python
import numpy as np
import os

# Path to the folder
folder_path = 'all_files'

# Get all file names
all_files = os.listdir(folder_path)

# Separate and sort input and invest files
input_files = sorted([f for f in all_files if f.startswith("RL_inputs_") and f.endswith(".npy")])
invest_files = sorted([f for f in all_files if f.startswith("RL_ordinal_invest_") and f.endswith(".npy")])

# Load and stack inputs
input_arrays = [np.load(os.path.join(folder_path, f)) for f in input_files]
stacked_inputs = np.stack(input_arrays, axis=0)

# Load and stack invest data
invest_arrays = [np.load(os.path.join(folder_path, f)) for f in invest_files]
stacked_invests = np.stack(invest_arrays, axis=0).reshape(32,-1,1)

# Optional: show shapes to verify
print("Stacked inputs shape:", stacked_inputs.shape)
print("Stacked invests shape:", stacked_invests.shape)
```

```python
np.save("stacked_inputs", stacked_inputs)
np.save("stacked_invests", stacked_invests)
```
